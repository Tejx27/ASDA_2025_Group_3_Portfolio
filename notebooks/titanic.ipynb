```python
#getting statistics on categorical/object columns
obj_cols = df.select_dtypes(include='object').columns
print(f"This dataset has {len(obj_cols)} object columns:", list(obj_cols))
```

```python
#detailed statistics on object data
for col in df.select_dtypes(include='object'):
    print(f"\nColumn: {col}")
    print("Count of non-null values:", df[col].count()) 
    print("Number of unique values:", df[col].nunique())
    print("Most frequent value:", df[col].value_counts().idxmax())
    print("Most frequent value (frequency):", df[col].value_counts().max())
    print("Least frequent value:", df[col].value_counts().idxmin())
    print("Least frequent value (frequency):", df[col].value_counts().min())
```

### 4. Data Consistency

    background-color: inherit;

```python
# Are there mixed value types in a column?
for col in df.columns:
    types = df[col].apply(type).value_counts()
    if len(types) > 1:
        print(f"{col}: mixed types -> {types.index.tolist()}")
    else:
        print(f"{col}: No mixed types")
```

```python
# Are string column clean?
for col in df.select_dtypes(include='object').columns:
    errors = df[col].str.contains(r'[^a-zA-Z\s]', na=False).sum()
    if errors > 0:
        print(f"{col}: {errors} rows with non-alphabetic characters")
    else:
        print(f"{col}: No errors")
```
