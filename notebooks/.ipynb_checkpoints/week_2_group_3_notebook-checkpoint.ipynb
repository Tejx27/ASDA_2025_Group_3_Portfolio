```python
# ASDA week 2 - dataset 1
import pandas as pd
```

```python
# importing data from .csv
column_names = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV']
housing_data = pd.read_csv('/Users/tejas/Desktop/Data Science/ASDA/Group/ASDA_2025_Group_3_Portfolio/additional_materials/housing.csv', header=None, delimiter=r"\s+", names=column_names)
housing_data.head()
```

```python
# number of (rows, columns)
housing_data.shape
```

```python
# Structure of dataset
housing_data.info()
```

```python
# unique values in columns
housing_data.nunique()
```

```python
# Descriptive statastucs => Numeric columns
statistics = pd.DataFrame({
    "Count": housing_data.count(),
    "Mean": housing_data.mean(),
    "Standard Deviation": housing_data.std(),
    "Min": housing_data.min(),
    "25%": housing_data.quantile(0.25),
    "50%": housing_data.quantile(0.5),
    "75%": housing_data.quantile(0.75),
    "Max": housing_data.max()
})
statistics

# Transpose the DataFrame to swap rows and columns
statistics_transposed = statistics.T
# Results
statistics_transposed
```

```python
# Categorical / Object Columns
categorical_cols = ["CHAS", "RAD"]

categorical = pd.DataFrame({
    "Count": housing_data[categorical_cols].count(),
    "Number of Unique Values": housing_data[categorical_cols].nunique(),
    "Most Frequent Value": housing_data[categorical_cols].mode().iloc[0],
    "Most Frequent Value (Frequency)": [
        housing_data[col].value_counts().iloc[0] for col in categorical_cols
    ],
    "Least Frequent Value": [
        housing_data[col].value_counts().index[-1] for col in categorical_cols
    ],
    "Least Frequent Value (Frequency)": [
        housing_data[col].value_counts().iloc[-1] for col in categorical_cols
    ]
})

categorical

# Transpose the DataFrame to swap rows and columns
categorical_transposed = categorical.T
# results
categorical_transposed
```

```python
# missing values and duplicates 
housing_data_null_duplicates = pd.DataFrame({
    "missing count" : housing_data.isnull().sum(),
    "% Missing" : housing_data.isnull().values.any()
})
housing_data_null_duplicates
```

```python
# missing values and percentage of missing values 
total_missing = housing_data.isnull().sum().sum()
percent_missing = (total_missing / housing_data.size) * 100
print(total_missing)
print(percent_missing)
```

```python
# Duplicate values and percentage of duplicate values 
duplicated_rows = housing_data.duplicated().sum()
percent_duplicates = (duplicated_rows / len(housing_data)) * 100
print(duplicated_rows)
print(percent_duplicates)
```
